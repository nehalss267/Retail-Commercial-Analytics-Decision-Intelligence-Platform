# 01 — Data Quality
Load and assess the UCI Online Retail dataset for quality issues.

```python
import pandas as pd
import numpy as np
from pathlib import Path

RAW = Path('..') / 'data' / 'raw' / 'Online Retail.xlsx'
print(f'Loading {RAW}...')
df = pd.read_excel(RAW, engine='openpyxl')
print(f'Shape: {df.shape}')
df.head()
```

## Schema & Types

```python
df.info()
print(f'\nShape: {df.shape}')
print(f'Columns: {list(df.columns)}')
```

## Missing Values

```python
missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)
pd.DataFrame({'count': missing, 'pct': missing_pct})[missing > 0]
```

## Duplicates

```python
n_dupes = df.duplicated().sum()
print(f'Exact duplicate rows: {n_dupes} ({n_dupes/len(df)*100:.2f}%)')
```

## Cancellations & Returns

```python
cancelled = df[df['InvoiceNo'].astype(str).str.startswith('C')]
print(f'Cancellations: {len(cancelled)} rows ({len(cancelled)/len(df)*100:.2f}%)')
print(f'Negative quantities: {(df["Quantity"] < 0).sum()} rows')
print(f'Negative prices: {(df["UnitPrice"] < 0).sum()} rows')
print(f'Zero prices: {(df["UnitPrice"] == 0).sum()} rows')
```

## Date Range & Coverage

```python
print(f'Date range: {df["InvoiceDate"].min()} to {df["InvoiceDate"].max()}')
print(f'Unique customers: {df["CustomerID"].nunique():,}')
print(f'Unique products: {df["StockCode"].nunique():,}')
print(f'Countries: {df["Country"].nunique()}')
```

## Country Distribution

```python
country_counts = df['Country'].value_counts()
print(f'UK share: {country_counts.iloc[0]/len(df)*100:.1f}%')
country_counts.head(10)
```

## Summary

```python
print('=== DATA QUALITY SUMMARY ===')
print(f'Rows: {len(df):,}')
print(f'Columns: {df.shape[1]}')
print(f'Missing CustomerID: {df["CustomerID"].isnull().sum():,} ({df["CustomerID"].isnull().mean()*100:.1f}%)')
print(f'Duplicates: {n_dupes:,}')
print(f'Cancellations: {len(cancelled):,}')
```
