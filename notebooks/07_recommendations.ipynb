# 07 — Recommendation System
Popularity, content-based, and collaborative filtering approaches.

```python
import json
import pandas as pd
from pathlib import Path

PROCESSED = Path('..') / 'data' / 'processed'
recs = json.loads((PROCESSED / 'recommendations_report.json').read_text())
recs.keys()
```

## 1. Popularity Baseline

```python
pop = pd.DataFrame(recs['popularity_top_20'])
print(f'Top products: {len(pop)}')
pop[['StockCode', 'Description', 'revenue', 'quantity', 'orders']].head(10)
```

## 2. Content-Based Recommendations

```python
content = recs['content_based_for_product']
print(f'Recommendations for product: {content["product"]}')
content_recs = pd.DataFrame(content['recommendations'])
content_recs
```

## 3. Collaborative Filtering

```python
collab = recs['collaborative_for_customer']
print(f'Recommendations for customer: {collab["customer_id"]}')
collab_recs = pd.DataFrame(collab['recommendations'])
collab_recs
```

## Evaluation Metrics

```python
from src.models.recommendation.evaluation import (
    precision_at_k, recall_at_k, ndcg_at_k
)

# Example evaluation
purchased = set(pop['StockCode'].head(5).tolist())
recommended = pop['StockCode'].head(10).tolist()

print(f'Precision@5: {precision_at_k(recommended, purchased, k=5):.2f}')
print(f'Recall@5:    {recall_at_k(recommended, purchased, k=5):.2f}')
print(f'NDCG@5:      {ndcg_at_k(recommended, purchased, k=5):.2f}')
```

## Methodology

- **Popularity**: Top products ranked by total revenue
- **Content-based**: Cosine similarity on product-customer revenue matrix
- **Collaborative filtering**: User-based cosine similarity on user-item matrix, top-20 similar users
