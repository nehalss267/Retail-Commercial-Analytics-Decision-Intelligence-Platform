# 04 — Statistical Analysis
Hypothesis tests, confidence intervals, and effect sizes.

```python
import json
from pathlib import Path

PROCESSED = Path('..') / 'data' / 'processed'
report = json.loads((PROCESSED / 'statistics_report.json').read_text())
report
```

## Test 1: High vs Low Value Customers (Welch's t-test)

```python
test = report['high_vs_low_value']
print(f'Test: {test["test"]}')
print(f'Significant: {test["significant"]}')
print(f'p-value: {test["p_value"]}')
print(f'Effect size (Cohen d): {test["effect_size_cohens_d"]}')
print(f'High value mean: £{test["high_value_mean"]}')
print(f'Low value mean: £{test["low_value_mean"]}')
```

## Test 2: Country AOV (ANOVA)

```python
test = report['country_aov']
print(f'Test: {test["test"]}')
print(f'Significant: {test["significant"]}')
print(f'F-statistic: {test["f_statistic"]}')
print(f'p-value: {test["p_value"]}')
print(f'Kruskal-Wallis H: {test["kruskal_h"]} (p={test["kruskal_p"]})')
print(f'Countries: {test["countries"]}')
```

## Test 3: Segment Differences (ANOVA)

```python
test = report['segment_differences']
print(f'Test: {test["test"]}')
print(f'Significant: {test["significant"]}')
print(f'F-statistic: {test["f_statistic"]}')
print(f'p-value: {test["p_value"]}')

import pandas as pd
pd.DataFrame(test['segment_stats'])
```

## Business Interpretation

- **All tests are statistically significant** (p < 0.05)
- High-value customers have significantly different order values from low-value customers
- AOV differs across countries — international markets behave differently
- Customer segments have distinct monetary profiles — segmentation is meaningful
